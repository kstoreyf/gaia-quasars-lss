```python
import numpy as np
from astropy.table import Table

%load_ext autoreload
%autoreload 2
```

# Load Quaia

Quaia can be downloaded here: https://zenodo.org/records/10403370

```python
G_lim = 20.5
tab_quaia = Table.read(f'../data/quaia_G{G_lim}.fits')
```

```python
tab_quaia.columns
```

# Print info

```python
z_min = 3.15
z_max = 3.89
tab_quaia_highz = tab_quaia[(tab_quaia['redshift_quaia']>z_min) & (tab_quaia['redshift_quaia']<z_max)]
print(f'{len(tab_quaia_highz)} quaia sources have {z_min}<z<{z_max}')
```

```python
z_min = 4
z_max = np.max(tab_quaia['redshift_quaia'])
tab_quaia_highz = tab_quaia[(tab_quaia['redshift_quaia']>z_min) & (tab_quaia['redshift_quaia']<z_max)]
print(f'{len(tab_quaia_highz)} quaia sources have {z_min:.3f}<z<{z_max:.3f}')
```

```python
G_min = np.min(tab_quaia['phot_g_mean_mag'])
G_max = 18
tab_quaia_Gbin = tab_quaia[(tab_quaia['phot_g_mean_mag']>G_min) & (tab_quaia['phot_g_mean_mag']<G_max)]
print(f'{len(tab_quaia_Gbin)} quaia sources have {G_min:.3f}<G<{G_max:.3f}')
```

```python
G_min = np.min(tab_quaia['phot_g_mean_mag'])
G_max = 19
tab_quaia_Gbin = tab_quaia[(tab_quaia['phot_g_mean_mag']>G_min) & (tab_quaia['phot_g_mean_mag']<G_max)]
print(f'{len(tab_quaia_Gbin)} quaia sources have {G_min:.3f}<G<{G_max:.3f}')
```

# Save light numpy version

```python
tab_quaia_lite = tab_quaia['source_id', 'l', 'b', 'phot_g_mean_mag', 'redshift_quaia']
tab_quaia_lite['source_id'] = tab_quaia_lite['source_id'].filled()
np.save('../data/quaia_forvia.npy', tab_quaia_lite.as_array())
```
